# 06 Model Ready Dataset

This notebook creates the final dataset artifacts meant to be consumed later. It intentionally saves only two CSV files to `../processed`:

- `training_model_ready.csv`
- `testing_model_ready.csv`

Train/validation splits should be created later inside a model-training notebook or service-specific training script.

```python
from pathlib import Path
import pandas as pd
```

```python
DATASET_DIR = Path("..").resolve()
INTERIM_DIR = DATASET_DIR / "interim"
PROCESSED_DIR = DATASET_DIR / "processed"
REPORTS_DIR = DATASET_DIR / "reports"

for directory in [PROCESSED_DIR, REPORTS_DIR]:
    directory.mkdir(parents=True, exist_ok=True)

TRAIN_FEATURES_PATH = INTERIM_DIR / "training_features.csv"
TEST_FEATURES_PATH = INTERIM_DIR / "testing_features.csv"
```

```python
train_df = pd.read_csv(TRAIN_FEATURES_PATH)
test_df = pd.read_csv(TEST_FEATURES_PATH)

target_col = "SeriousDlqin2yrs"

print(train_df.shape, test_df.shape)
```

```python
training_model_ready = train_df[train_df[target_col].notna()].copy()
training_model_ready[target_col] = training_model_ready[target_col].astype(int)

testing_model_ready = test_df.drop(columns=[target_col], errors="ignore").copy()

print("training_model_ready:", training_model_ready.shape)
print("testing_model_ready:", testing_model_ready.shape)
```

```python
print("Training missing values:", training_model_ready.isna().sum().sum())
print("Testing missing values:", testing_model_ready.isna().sum().sum())

training_model_ready.head()
```

```python
training_model_ready.to_csv(PROCESSED_DIR / "training_model_ready.csv", index=False)
testing_model_ready.to_csv(PROCESSED_DIR / "testing_model_ready.csv", index=False)

report_text = f"""# Model Ready Dataset

## Final CSV Outputs

- `datasets/processed/training_model_ready.csv`
- `datasets/processed/testing_model_ready.csv`

## Shapes

| Dataset | Rows | Columns |
|---|---:|---:|
| Training model ready | {training_model_ready.shape[0]:,} | {training_model_ready.shape[1]:,} |
| Testing model ready | {testing_model_ready.shape[0]:,} | {testing_model_ready.shape[1]:,} |

## Notes

The processed folder intentionally contains final model-ready datasets only. Train/validation splits should be produced later in the model-training workflow.
"""

(REPORTS_DIR / "model_ready_dataset.md").write_text(report_text, encoding="utf-8")
print("Saved final model-ready datasets and report.")
```
